### 4. data cleaning

```python
df_clean = df.drop(columns=['id','age','sleep_hours','notifications_per_day','app_opens_per_day'])
df_clean.isna().sum()
```

```python
from scipy.stats import chi2_contingency

alpha = 0.05
chi2_results = {}
cat_features = ['gender','stress_level','academic_work_impact']
for feature in cat_features:
  contingency_table = pd.crosstab(df_clean[feature], df_clean['addicted_label'])
  chi2, p, dof, expected = chi2_contingency(contingency_table)
  decision = 'Reject Null (keep feature)' if p < alpha else 'Fail to Reject Null (drop feature)'
  chi2_results[feature] = {
      'Chi-squared Statistic': chi2,
      'p-value': p,
      'Decision': decision
  }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df.sort_values(by='p-value', ascending=True)
```

### 5. train test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df_clean.drop(columns=['addicted_label']), df_clean['addicted_label'], test_size=0.3, random_state=42, stratify=df_clean['addicted_label'])
```

### 6. pipeline and column transformer

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder, FunctionTransformer, OrdinalEncoder
from sklearn.compose import ColumnTransformer
```

```python
skewed_cols = ['social_media_hours','gaming_hours','work_study_hours']
plain_numeric_cols = ['daily_screen_time_hours','weekend_screen_time']
ordinal_col = ['stress_level']
categorical_cols = ['gender','academic_work_impact']
```

```python
skewed_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("log_transform", FunctionTransformer(np.log1p)),
    ("scaler", StandardScaler())
])

plain_numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

ordinal_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encode", OrdinalEncoder(categories=[['Low', 'Medium', 'High']]))
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encode", OneHotEncoder(handle_unknown="ignore",
    drop="if_binary"))
])


preprocessor = ColumnTransformer(transformers=[
    ("skewed", skewed_pipeline, skewed_cols),
    ("plain_numeric", plain_numeric_pipeline, plain_numeric_cols),
    ("ordinal", ordinal_pipeline, ordinal_col),
    ("categorical", categorical_pipeline, categorical_cols)
])
```

```python
preprocessor
```

### 7. trying multiple algorithms

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix

from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "f1_macro": "f1_macro"
}
```

```python
models = {
    'Logistic Regression': LogisticRegression(class_weight='balanced',random_state=42)
    , 'Decision Tree': DecisionTreeClassifier(class_weight='balanced',random_state=42)
    , 'Gaussian Naive Bayes': GaussianNB()
    # , 'Support Vector Machine': SVC(class_weight='balanced',random_state=42)
    , 'KNN': KNeighborsClassifier()
}

for name, model in models.items():
    pipe = Pipeline(steps=[('preprocessor', preprocessor)
    , ('classifier', model)])

    if name in ['Support Vector Machine', 'KNN']:
      results = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=StratifiedKFold(
            n_splits=2,
            shuffle=True,
            random_state=42
        ),
        scoring=scoring,
        n_jobs=-1
      )
    else:
      results = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
      )

    print(f"\n{name}")
    print(f"Accuracy:           {results['test_accuracy'].mean():.3f}")
    print(f"Balanced accuracy:  {results['test_balanced_accuracy'].mean():.3f}")
    print(f"Macro F1:           {results['test_f1_macro'].mean():.3f}")
```

### 8. trying ensemble learning

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
```

```python
models = {
    'Random Forest': RandomForestClassifier(class_weight='balanced',random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42),
    'XGBoost': XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
}

for name, model in models.items():
    pipe = Pipeline(steps=[('preprocessor', preprocessor)
    , ('classifier', model)])

    results = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    print(f"\n{name}")
    print(f"Accuracy:           {results['test_accuracy'].mean():.3f}")
    print(f"Balanced accuracy:  {results['test_balanced_accuracy'].mean():.3f}")
    print(f"Macro F1:           {results['test_f1_macro'].mean():.3f}")
```

### 9. final evaluation on X_test

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score

best_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42))
])

results = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

print(f"\n{name}")
print(f"Accuracy:           {results['test_accuracy'].mean():.3f}")
print(f"Balanced accuracy:  {results['test_balanced_accuracy'].mean():.3f}")
print(f"Macro F1:           {results['test_f1_macro'].mean():.3f}")

best_pipeline.fit(X_train, y_train)

y_pred = best_pipeline.predict(X_test)

print(f'Accuracy Score: {accuracy_score(y_test, y_pred)}')
print(f'F1 Score: {f1_score(y_test, y_pred, average="macro")}')
print(f'Confusion Matrix:\n {confusion_matrix(y_test, y_pred)}')
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred),
            annot=True,
            fmt='.2f',
            xticklabels=best_pipeline.classes_,
            yticklabels=best_pipeline.classes_)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
